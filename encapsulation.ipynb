```python
class employee:

    def __init__(self,name,salary):
        self.name = name            #public attribute
        self.__salary = salary      #private attribute


emp = employee("mishwa",100)
print(emp.name)    
print(emp.__salary)
```

```python
class person:

    def __init__(self,name,age,address):
        self.name = name
        self.age = age
        self._address = address

class subpersonal(person):

    def showdetails(self):
        print("address:-",self.__address)
        
p = subpersonal("Mishwa",22,"kamrej")
print(p.name)
print(p.age)
p.showdetails()


            
            

     
```
