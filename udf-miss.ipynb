```python
# 13. Function to count even and odd numbers in a list

def even_odd(no):

    even = 0
    odd = 0

    for number in no:

        if number % 2 == 0:
            even += 1
        else:
            odd +=1

    return even, odd

no=[10,2,5,7,50,47,77,36,12,20]
even, odd = even_odd(no)

print("Even No:-",even)
print("Odd No:-",odd)
```
