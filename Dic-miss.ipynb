```python
#11. Create a dictionary of 5 students and their marks

students={
    "priyanka":90,
    "pratik":85,
    "rahul":75,
    "priya":95,
    "ajay":80
}
print(students)
```

```python
#12. Find the student with the highest marks

students={
    "priyanka":95,
    "pratik":85,
    "rahul":75,
    "priya":90,
    "ajay":80
}

highest_marks=0
highest_student=""
for student, marks in students.items():
    if marks>highest_marks:
        highest_marks=marks
        highest_student=student
print("student:",highest_student)
print("marks:",highest_marks)
```

```python
#13. Find the student with the lowest marks

students={
    "priyanka":95,
    "pratik":85,
    "rahul":23,
    "priya":90,
    "ajay":80
}

lowest_marks=100
lowest_student=""
for student, marks in students.items():
    if marks<lowest_marks:
        lowest_marks=marks
        lowest_student=student
print("student:",lowest_student)
print("marks:",lowest_marks)
```

```python
#14. Count how many students passed and failed

students={
    "priyanka":90,
    "pratik":85,
    "rahul":32,
    "priya":23,
    "ajay":80
}
passed=0
failed=0
for marks in students.values():
    if marks>=35:
        passed+=1
    else:
        failed+=1
print("students passed:", passed)
print("students failed:", failed)
```
