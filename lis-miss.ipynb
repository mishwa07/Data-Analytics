**14. Count Even and Odd Numbers**

```python
no=[10,15,20,25,30,35,40,45,50]

even=0
odd=0

for numbers in no:
    if numbers % 2 == 0:
        even+=1
    else:
        odd+=1

print("Even No:", even)
print("Odd No:", odd)
```

**18. Find Second Largest**

```python
no = [25, 10, 45, 30, 15]

largest = no[0]
second_largest = no[0]

for numbers in no:
    if numbers > largest:
        second_largest = largest
        largest = numbers
    elif numbers > second_largest:
        second_largest = numbers

print("Second Largest:", second_largest)
```

**27. Move Zeros to the End**

```python
numbers = [0, 10, 0, 20, 30, 0, 40]
result = []
zero_count = 0
for number in numbers:
    if number == 0:
        zero_count += 1
    else:
        result.append(number)
for i in range(zero_count):
  result.append(0)
print(result)
```
